### 2. Prompts

- Prompts are the input give to LLM to guide its output



- Prompts can be text based or multimodal like image or audio or video
- They are very sensitive becasue very slight change in prompt will change the output
- Ideally in any app we develop, user will be passing the **static** input prompt which will be passed to LLM and get the output but since input prompt is very sensitive so we will not have any control over the output, model may halucinate and may not give the correct output or the consistent output which we want to provide to the user --> In order to control the output, we can use **dynamic prompts** using **langchain's prompt_template** to control the output.

### 3. Structured output in Langchain

- By default, LLMs return responses in free-form, unstructured text. **Structured Output** in LangChain ensures that responses are generated in a predefined format such as JSON, making them easier to parse and use programmatically.

- It will be easier for
    1. data extarction
    2. API building
    3. AI Agents etc



- There are 2 type of LLMs
    1. LLMs which can return strucutred output
    2. LLMs which can not return structured output

#### 1. With structured output function

- Before model invoke we need to call ith strucutred output function and provide data format
- There are 3 types to specify data format
    1. Typed dict
    2. Pydantic
    3. Json_Schema

#### 1. Typedict

- **Typed dict**
    - Its a way to define a dictionary in python where you specify what keys and values should exist
    - It helps to ensure that dictionary follows a specific structure
    - It tells what keys are required and what type of values they should have
    - It doesnt validate, It just tells us but even if you provide a wrong dtype then will not throw error

    
```
from langchain_openai import ChatOpenAI
from dotenv import load_dotenv;load_dotenv()
from typing import TypedDict, Annotated

model = ChatOpenAI(
    model="gpt-4o-mini",
    temperature=0,
    max_completion_tokens=100
)

# Schema
class Review(TypedDict):
    summary: str
    sentiment: str
    
    # float score between -1 and 1
    sentiment_score: Annotated[
        float,
        "Sentiment score ranging from -1 (very negative) to 1 (very positive)"
    ]

structured_model = model.with_structured_output(Review)

result = structured_model.invoke(
    "Great service, friendly staff, and excellent experience overall. Highly recommended!"
)

print(result)

{'summary': 'Great service, friendly staff, and excellent experience overall. Highly recommended!', 'sentiment': 'positive', 'sentiment_score': 0.95}
```


- When we use structured output, In backend it generates a prompt to return output in same strucutred schema without even asking anything about sentiment, sentiment score or summary since that's how the structured schema was defined it returns the output in the same way

- Use **Annotated** to define the exact output needed from LLM without it getting hallucinated


```
from langchain_openai import ChatOpenAI
```

#### 2. Pydantic

- We can just give input which will trigger all component and finally we will get the output

```
### How It Works
- In a two-class problem, SVM identifies the hyperplane that maximizes the **margin** – the distance between the hyperplane and the nearest data points from each class.
- These closest points are called **support vectors**. They alone define the decision boundary.
- If data is not linearly separable, SVM uses **kernel functions** (e.g., linear, polynomial, RBF) to map data into a higher dimension where a clear separation is possible.

### Key Features
- **Effective in high dimensions** – works well even when number of features exceeds samples.
- **Memory efficient** – only support vectors are retained after training.
- **Versatile** – different kernels allow handling of complex, non-linear boundaries.
- **Sensitive to scaling** – feature normalization is essential.
- **Not ideal for very large datasets** – training time can be high.

### Common Applications
Text classification, image recognition, bioinformatics (e.g., protein classification), and handwriting recognition.

In short, SVM is a powerful, geometry-driven classifier known for its accuracy and robustness, especially in moderate-sized, high-dimensional problems."""
```

```
    sentiment: Literal['positive', 'negative'] = Field(description="give the sentiment of the feedback")
```

### 6. Runnables

#### Why

- Chain in Langchain enables three core execution flows:
    1. **Sequential** – step-by-step chain execution  
    2. **Parallel** – concurrent runs of multiple chains  
    3. **Conditional** – logic-based branching  

- Efficient chaining requires a solid grasp of **runnables**.  

- When ChatGPT launched its public APIs in 2022, businesses rapidly adopted them to build chatbots.  
Recognizing the growing need for LLM-based apps, LangChain created a unified framework that supports any LLM with minimal code changes.  

- LangChain gained popularity because LLM interaction is just the start—it also handles **data loading, embeddings, and semantic search** seamlessly.



- LangChain expanded beyond core execution by building helper functions for each component, making it a powerful library.

**Prompt** is common across all LLM apps — this led to the introduction of **Chain**.

- For RAG-based applications, **retrieval** is a recurring need. LangChain integrated all retrieval stages into the **Retriever QA Chain**.

- LangChain has since developed multiple specialized chains for common tasks.



- Over time, LangChain introduced many chains, increasing complexity for new learners.

- Components were developed independently and lacked standardization, leading to inconsistent behavior. This forced the team to create custom chain functions like `LLMChain`, `PromptChain`, and `QAChain`.

- To enable seamless component integration, LangChain introduced **Runnables** — a unified interface for building and connecting chains consistently.

#### What



- Runnables are a unit of work, which means:
    1. We can give an **input**
    2. It will **process**
    3. Returns an **output**

- Each runnable follows a common interface
    1. invoke
    2. batch
    3. stream

- Each runnable can be connected together, which makes output of 1st runnable becomes input for next runnable

- Each runnble workflow is itself a runnable

- Runnables are same as **Lego blocks**

````python
from abc import ABC, abstractmethod
````

````python
class Runnable(ABC):

  @abstractmethod
  def invoke(input_data):
    pass
````

````python
import random

class NakliLLM(Runnable):

  def __init__(self):
    print('LLM created')

  def invoke(self, prompt):
    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}


  def predict(self, prompt):

    response_list = [
        'Delhi is the capital of India',
        'IPL is a cricket league',
        'AI stands for Artificial Intelligence'
    ]

    return {'response': random.choice(response_list)}
````

````python
class NakliPromptTemplate(Runnable):

  def __init__(self, template, input_variables):
    self.template = template
    self.input_variables = input_variables

  def invoke(self, input_dict):
    return self.template.format(**input_dict)

  def format(self, input_dict):
    return self.template.format(**input_dict)
````

````python
class NakliStrOutputParser(Runnable):

  def __init__(self):
    pass

  def invoke(self, input_data):
    return input_data['response']
````

````python
class RunnableConnector(Runnable):

  def __init__(self, runnable_list):
    self.runnable_list = runnable_list

  def invoke(self, input_data):

    for runnable in self.runnable_list:
      input_data = runnable.invoke(input_data)

    return input_data
````
